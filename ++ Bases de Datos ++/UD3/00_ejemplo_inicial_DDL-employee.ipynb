# ¿Qué es una Biblioteca (Librería)?
## Definición de Biblioteca (Librería)
Biblioteca (Librería): En programación, una biblioteca es un conjunto de códigos y funciones preescritas que puedes usar para hacer tu trabajo más fácil. Piensa en ello como una caja de herramientas donde tienes herramientas ya hechas para realizar tareas específicas sin tener que construirlas desde cero.

## ¿Por qué Usamos Bibliotecas?
- **Reutilización de Código**: Te permite usar código que otros programadores ya han escrito y probado, lo que ahorra tiempo y reduce errores.
- **Facilidad de Uso**: Proporciona funciones y métodos que simplifican tareas complejas. Por ejemplo, en vez de escribir todo el código para conectarte a una base de datos, puedes usar una biblioteca que lo haga por ti con solo unas pocas líneas de código.

# ¿Por qué usamos SQLite en Memoria?
## ¿Qué es SQLite?
SQLite es una biblioteca de software que proporciona una base de datos SQL sin servidor, autocontenida y de alta confiabilidad. Imagina que es como una hoja de cálculo muy avanzada donde puedes almacenar, organizar y consultar datos de forma eficiente.

A diferencia de otras bases de datos más grandes que necesitan configuraciones complicadas y servidores especiales para funcionar, SQLite es muy ligera y se puede usar directamente en el programa.

## ¿Por qué en Memoria?
- **Base de Datos en Memoria**: Cuando decimos que usamos SQLite en memoria, significa que toda la base de datos se guarda temporalmente en la memoria del ordenador en lugar de en el disco duro. Esto hace que sea muy rápida, pero los datos desaparecen cuando apagas el ordenador o cierras el programa. Cuando usamos SQLite en memoria, estamos creando una base de datos nueva cada vez que ejecutamos nuestro programa. Esto es útil para pruebas y aprendizaje porque no tenemos que preocuparnos por mantener y limpiar los datos entre sesiones.
  
### Ventajas:
- **Velocidad**: Las operaciones son más rápidas porque los datos están en la memoria.
- **Pruebas y Desarrollo**: Es ideal para aprender, hacer pruebas y desarrollar proyectos pequeños sin necesidad de configurar un entorno complejo.

# Explicación de Sentencias SQL en Python
## Sentencia 1: `conn = sqlite3.connect(':memory:')`

### ¿Qué hace esta sentencia?
- **Propósito**: Esta línea de código crea una conexión a una base de datos SQLite que se almacena en la memoria.
- **Analogía**: Imagina que estás abriendo un nuevo cuaderno para escribir notas. Pero en lugar de guardar el cuaderno en tu escritorio (el disco duro), estás escribiendo las notas en una pizarra temporal (la memoria). Esto significa que las notas solo existirán mientras la pizarra esté abierta y desaparecerán cuando la borres.

### ¿Por qué es Necesaria?
- **Crear una Base de Datos Temporal**: Al usar `':memory:'`, estamos diciendo que queremos una base de datos que solo viva en la memoria mientras el programa está corriendo. Es ideal para pruebas y desarrollo porque es rápida y no deja datos residuales.
- **Conexión**: Esta línea establece una conexión a la base de datos que usaremos para enviar y recibir datos. Sin esta conexión, no podríamos interactuar con la base de datos.

## Sentencia 2: `cursor = conn.cursor()`

### ¿Qué hace esta sentencia?
- **Propósito**: Crea un cursor, que es un objeto que permite ejecutar sentencias SQL y gestionar los resultados.
- **Analogía**: Siguiendo con la analogía del cuaderno, si `conn` es como abrir un nuevo cuaderno, el `cursor` es como sacar un bolígrafo para escribir en ese cuaderno. Necesitamos el bolígrafo (cursor) para poder escribir y leer notas en el cuaderno (base de datos).

### ¿Por qué es Necesaria?
- **Ejecución de Sentencias SQL**: El cursor es el objeto que usamos para ejecutar nuestras sentencias SQL, como `CREATE TABLE`, `INSERT`, `SELECT`, etc. Sin el cursor, no tendríamos una forma de interactuar con la base de datos.
- **Gestión de Resultados**: El cursor nos permite no solo ejecutar las sentencias, sino también recuperar los resultados de las consultas que hagamos.

---

Algunos de los tipos de datos que maneja SQLite son:

1. **TEXT**: Es el tipo de datos utilizado para almacenar cadenas de texto de longitud variable. Es equivalente a `VARCHAR` en otros sistemas de bases de datos.

2. **INTEGER**: Es el tipo de datos utilizado para almacenar números enteros, ya sean enteros de 32 bits o 64 bits dependiendo de la plataforma.

3. **REAL**: Es el tipo de datos utilizado para almacenar números de punto flotante, equivalentes a `FLOAT` o `DOUBLE` en otros sistemas.

4. **BLOB**: Es el tipo de datos utilizado para almacenar datos binarios, como archivos o imágenes.

5. **date**: Es el tipo de datos utilizado para almacenar fechas.


Por otro lado, en sistemas como Oracle encontrarás `VARCHAR2` y `NUMBER` como tipos de datos específicos, en lugar de utilizar `TEXT`, `INTEGER`, y `REAL` para almacenar datos de texto y números, respectivamente.
